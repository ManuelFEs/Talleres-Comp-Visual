```python
import cv2
import numpy as np
import imageio
from matplotlib import pyplot as plt
```

```python
img = cv2.imread('./data/redPanda.jpg')
img = cv2.cvtColor(img, cv2.COLOR_RGBA2BGR)
```

```python
img
```

```python
b, g, r = cv2.split(img)
red_channel_img = cv2.merge([np.zeros_like(b), np.zeros_like(g), r])
red_channel_img = cv2.cvtColor(red_channel_img, cv2.COLOR_BGR2RGB)
red_channel_img = cv2.cvtColor(red_channel_img, cv2.COLOR_BGR2GRAY)

blue_channel_img = cv2.merge([b, np.zeros_like(g), np.zeros_like(r)])
blue_channel_img = cv2.cvtColor(blue_channel_img, cv2.COLOR_BGR2RGB)
blue_channel_img = cv2.cvtColor(blue_channel_img, cv2.COLOR_BGR2GRAY)

green_channel_img = cv2.merge([np.zeros_like(b), g, np.zeros_like(r)])
green_channel_img = cv2.cvtColor(green_channel_img, cv2.COLOR_BGR2RGB)
green_channel_img = cv2.cvtColor(green_channel_img, cv2.COLOR_BGR2GRAY)
```

```python
red_channel_img, green_channel_img, blue_channel_img
```

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gray_img
```

```python
kernel = np.ones((5,5),np.uint8)
erosion = cv2.erode(gray_img,kernel,iterations = 1)
dilation = cv2.dilate(gray_img,kernel,iterations = 1)
opening = cv2.morphologyEx(gray_img, cv2.MORPH_OPEN, kernel, iterations=1)

opening
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

processed_frames = [
    img_rgb,
    cv2.cvtColor(red_channel_img, cv2.COLOR_GRAY2BGR),
    cv2.cvtColor(green_channel_img, cv2.COLOR_GRAY2BGR),
    cv2.cvtColor(blue_channel_img, cv2.COLOR_GRAY2BGR),
    gray_img, # Already 3-channel, so no further conversion needed
    erosion,  # Already 3-channel, so no further conversion needed
    dilation, # Already 3-channel, so no further conversion needed
    opening   # Already 3-channel, so no further conversion needed
]
print("Image frames prepared successfully.")
```

```python
import imageio

imageio.mimsave('image_changes.gif', processed_frames, duration=8.5)
print("GIF 'image_changes.gif' created successfully.")
```
